```python
model = transformer.variants.gemma.gemma_from_pretrained_checkpoint(flat_params, upcast_activations_to_float32=False)
# Block sublayer types should now be: Residual, Residual, PerLayerEmbeddingInject, LearnableRescale
print([type(s).__name__ for s in model.body.sublayers[3].sublayers])
# A KV-shared layer (e.g., the last block) should show this in its key path:
print([type(s).__name__ for s in model.body.sublayers[-3].sublayers[0].delta.sublayers[1].input_to_key.__class__.__bases__])
```

```python
model = pz.select(model).at_instances_of(pz.StateVariable).apply(lambda sv: pz.StateVariableSlot(label=sv.label))
```

```python
# Use the wrapped tokens
ntokens = tokens #pz.nx.wrap(tokens).tag("seq")
print("input:", type(ntokens).__name__, ntokens.named_shape, ntokens.positional_shape)

# Walk the body sublayers one-by-one and report what each returns
v = ntokens
for i, layer in enumerate(model.body.sublayers):
    try:
        v = layer(v, token_positions=pz.nx.arange("seq", ntokens.named_shape["seq"]),
                  token_ids=ntokens)
    except Exception as e:
        print(f"[{i}] {type(layer).__name__}: FAILED -> {type(e).__name__}: {e}")
        break
    print(f"[{i}] {type(layer).__name__:40s} -> "
          f"{type(v).__name__} named={dict(v.named_shape)} "
          f"pos={v.positional_shape}")
```

```python
# import jax
ntokens = tokens#pz.nx.wrap(tokens).tag("seq")
positions = pz.nx.arange("seq", ntokens.named_shape["seq"])

# Run body up to block 0
v = ntokens
for layer in model.body.sublayers[:3]:
    v = layer(v, token_positions=positions, token_ids=ntokens)

block0 = model.body.sublayers[3]
print("block0 type:", type(block0).__name__, "len(sublayers):", len(block0.sublayers))

for i, sub in enumerate(block0.sublayers):
    try:
        v = sub(v, token_positions=positions, token_ids=ntokens)
    except Exception as e:
        print(f"[block0.{i}] {type(sub).__name__}: FAILED -> {type(e).__name__}: {e}")
        # Walk into the failing residual's delta
        if hasattr(sub, "delta"):
            inner = sub.delta
            print(f"  delta type: {type(inner).__name__}")
            if hasattr(inner, "sublayers"):
                # Need fresh v -- re-run up to here
                v2 = ntokens
                for L in model.body.sublayers[:3]:
                    v2 = L(v2, token_positions=positions, token_ids=ntokens)
                for j, ss in enumerate(inner.sublayers):
                    try:
                        v2 = ss(v2, token_positions=positions, token_ids=ntokens)
                    except Exception as e2:
                        print(f"  [delta.{j}] {type(ss).__name__}: FAILED -> {e2}")
                        break
                    print(f"  [delta.{j}] {type(ss).__name__:30s} -> {type(v2).__name__} named={dict(v2.named_shape)} pos={v2.positional_shape}")
        break
    print(f"[block0.{i}] {type(sub).__name__:30s} -> {type(v).__name__} named={dict(v.named_shape)} pos={v.positional_shape}")
```
